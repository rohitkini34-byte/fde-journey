## 1. `if / elif / else`
Only the **first true branch** runs, so put the strictest condition first.

```python
score = 95
if score >= 50:
    print("pass")          # runs, and the chain stops here
elif score >= 90:
    print("distinction")   # never reached, even though it is true
```

```python
x = 7
print(0 <= x <= 12)        # chained comparison

items = []
# short-circuit: len(items) > 0 is False, so items[0] is never evaluated
print(len(items) > 0 and items[0] == "a")

value = None
print(value is None)       # use `is` only for None
```

## 2. Loops

```python
skills = ["Python", "SQL", "RAG"]
for i, skill in enumerate(skills, start=1):
    print(i, skill)

print(list(range(0, 10, 3)))   # stop (10) is excluded

total = 0                       # accumulator pattern
for n in [4, 5, 6]:
    total += n
print(total)
```

```python
# while: unknown number of repetitions (retries, pagination)
attempt = 0
while True:
    attempt += 1
    if attempt == 1:
        continue          # skip the rest of this round
    print("attempt", attempt)
    if attempt == 3:
        break
```

```python
# never remove from a list while looping over it -> build a new one
nums = [1, 2, 2, 3]
kept = []
for n in nums:
    if n != 2:
        kept.append(n)
print(kept)
```

## 3. Functions

```python
def area(width: float, height: float = 1.0) -> float:
    """Return the area of a rectangle."""
    if width < 0 or height < 0:
        return 0.0            # early return
    return width * height

print(area(3, 4), area(3), area(height=2, width=5))
```

## Re-test: the two misses (Q3 and Q4)

**Q3: return vs print.** A function with no `return` returns `None`.
- *What happens:* it prints `5`, then `None`.
- *Why:* `g` prints inside itself, but nothing is returned, so `result` is `None`.
- *Avoid it:* `return` the value and let the caller decide whether to print it.

```python
def g(x):
    print(x + 1)

result = g(4)
print(result)
```

**Q4: the mutable default trap.**
- *What happens:* the second call sees the first call's item.
- *Why:* the default `[]` is created **once**, when `def` runs, and every call that uses the default shares it.
- *Avoid it:* default to `None` and create a fresh list inside.

```python
def add_bad(item, bag=[]):
    bag.append(item)
    return bag

print(add_bad("a"), add_bad("b"))      # ['a', 'b'] ['a', 'b']  <- leaked

def add_good(item, bag=None):
    if bag is None:
        bag = []
    bag.append(item)
    return bag

print(add_good("a"), add_good("b"))    # ['a'] ['b']
```

**`is_prime`.** Why √n is enough: if `n = a × b` and both factors were bigger than √n, their product would be bigger than n.
So every composite number has a factor ≤ √n. Time O(√n).

```python
from dsa import is_prime, count_vowels

print([n for n in range(50) if is_prime(n)])
assert not is_prime(1) and not is_prime(91) and is_prime(97)
print(count_vowels("Forward Deployed Engineer"), count_vowels("rhythm"))
```
