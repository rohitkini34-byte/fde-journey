## 1. How Python runs
- Source → bytecode → Python virtual machine, line by line. An error only appears when its line is reached.
- Read tracebacks **bottom-up**: the last line is the error type and message, the lines above show where it happened.
- Script mode (`python3 file.py`) vs the REPL (`python3` then type lines).

```python
def broken():
    return 1 / 0

print("this line runs first")
try:
    broken()
except ZeroDivisionError as e:
    print("the error only happened when line 2 was reached:", e)
```

## 3. Variables are labels, not boxes

```python
a = [1, 2, 3]
b = a            # b is a second label on the SAME list
b.append(4)
print(a, a is b, id(a) == id(b))

c = a.copy()     # an independent list
c.append(5)
print(a, c, a is c)
```

## 4. Core types

```python
print(7 / 2, 7 // 2, 7 % 2)     # true division, floor division, remainder
print(0.1 + 0.2, 0.1 + 0.2 == 0.3)

from decimal import Decimal
print(Decimal("0.1") + Decimal("0.2"))   # exact
price_paise = 49_999                       # store money as integer paise
print(f"₹{price_paise / 100:.2f}")
```

## 5. Dynamic but strong typing

```python
try:
    "5" + 5
except TypeError as e:
    print("TypeError:", e)

print(int("5") + 5, "5" + str(5))
# input() ALWAYS returns a str -> convert with int() / float()
```
