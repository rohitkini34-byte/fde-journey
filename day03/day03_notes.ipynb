```python
scores = [72, 95, 60]
print(scores.sort(), scores)          # sort() returns None, sorts in place
scores = [72, 95, 60]
print(sorted(scores), scores)         # sorted() returns a new list
print(sorted(["b", "C", "a"], key=str.lower))
```

```python
import copy
grid = [[0, 0], [0, 0]]
shallow = grid.copy()
shallow[0][0] = 99
print(grid)                     # inner list is shared

grid = [[0, 0], [0, 0]]
deep = copy.deepcopy(grid)
deep[0][0] = 99
print(grid, deep)               # fully independent
```

## Concept 3: Dictionaries

```python
entry = {
    "uid": "blt123",
    "title": "Pricing page",
    "locale": "en-us",
    "tags": ["pricing", "landing"],
    "seo": {"meta_title": "Plans", "noindex": False},
}
print(entry["seo"]["meta_title"], entry["tags"][0])
print(entry.get("author"), entry.get("author", "Unknown"))

entry["status"] = "published"
del entry["locale"]
removed = entry.pop("tags")
entry.update({"version": 3, "status": "draft"})
print(entry, removed)
print("seo" in entry, "Plans" in entry)   # `in` checks keys only
```

```python
tickets = [{"id": 1, "priority": "high"}, {"id": 2, "priority": "low"}, {"id": 3, "priority": "high"}]

counts: dict[str, int] = {}
for t in tickets:
    counts[t["priority"]] = counts.get(t["priority"], 0) + 1

by_priority: dict[str, list[int]] = {}
for t in tickets:
    by_priority.setdefault(t["priority"], []).append(t["id"])

print(counts, by_priority)
```

## Concept 4: Sets

```python
in_cms = {"doc1", "doc2", "doc3", "doc4"}
in_index = {"doc2", "doc3", "doc5"}
print("missing from index:", in_cms - in_index)
print("stale in index:   ", in_index - in_cms)
print("both:             ", in_cms & in_index)
print("all:              ", in_cms | in_index)
print(type({}), type(set()))
print(list(dict.fromkeys([3, 1, 3, 2])))    # de-dupe and keep order
```

```python
# Big-O taste: list membership is O(n), set membership is about O(1)
import timeit
ids_list = list(range(100_000))
ids_set = set(ids_list)
print("list:", timeit.timeit(lambda: 99_999 in ids_list, number=200))
print("set: ", timeit.timeit(lambda: 99_999 in ids_set, number=200))
```

## Concept 6: Comprehensions

```python
hours = [5, 0, 3.5, 6, 0, 4]
print([h * 60 for h in hours if h > 0])                 # filter at the end
print(["long" if h >= 5 else "short" for h in hours])   # if/else before `for` picks a value

plans = {"basic": 999, "pro": 2999, "enterprise": 9999}
print({n: p for n, p in plans.items() if p < 5000})
print({e.lower() for e in ["A@x.com", "a@x.com", "b@y.com"]})

users = [{"id": 7, "name": "Asha"}, {"id": 9, "name": "Ravi"}]
by_id = {u["id"]: u for u in users}
print(by_id[9]["name"])
```

1. `nums = nums.sort()` → prints `None`. `sort()` sorts in place and returns `None`, so the name now points at `None`.
2. `{'x': [1, 2, 3]}`. `.copy()` is shallow: `b` is a new dict whose `"x"` value is the same list object, so appending through `b` changes `a`.
3. `False <class 'dict'>`. `{}` is an empty dict, not a set, and a dict never equals a set.
4. `n/a False`. `.get` returns the default for a missing key, and `in` checks keys, not values (`"Asha"` is a value).
5. `[0, 4, 16]`. Squares of the even numbers in 0–5.

## DSA
**Two Sum.** The brute force tries every pair: **O(n²)** time, O(1) space.
The single pass stores `value → index` in a dict. For each `n`, it asks whether `target - n` has been seen yet. That is an O(1) lookup, so the whole thing is **O(n)** time and O(n) space.

**First unique character.** Pass 1 counts every character in a dict, and pass 2 returns the first index whose count is 1. O(n) time. Space is O(k) for k distinct characters.

```python
from dsa import two_sum_brute, two_sum, first_unique_char

assert two_sum_brute([2, 7, 11, 15], 9) == two_sum([2, 7, 11, 15], 9) == (0, 1)
assert two_sum([3, 2, 4], 6) == (1, 2) and two_sum([3, 3], 6) == (0, 1)
assert first_unique_char("leetcode") == 0
assert first_unique_char("loveleetcode") == 2
assert first_unique_char("aabb") == -1
print("DSA checks passed")
```

## Quiz answers
1. **50,000 ticket IDs, already processed?** Use a `set` (or a dict keyed by ID). Membership is about O(1) through hashing, while a list is O(n) per check. 50,000 checks against a list would mean up to billions of comparisons.
2. **`entry["author"]` vs `entry.get("author")`.** `[]` raises `KeyError` when the key is missing, and `.get()` returns `None` or your default. Pick `[]` on purpose when the key *must* exist, so that bad data fails loudly and early instead of flowing on as `None`.
3. `result = [w.upper() for w in words if len(w) > 3]`
